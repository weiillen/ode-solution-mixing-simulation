```python
''' how different inflow/outflow rates (R) influence the quantity of sugar over time,
with fixed initial sugar quantity Q0 and inflow concentration. '''

import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# ODE function: dQ/dt = R * (X - Q/100)
def sugar_ode(t, Q, R, X):
    return R * (X - Q / 100)

# Solver function for the ODE
def solve_sugar_dynamics(Q0, X, R, t_max):
    t_span = (0, t_max)
    t_eval = np.linspace(0, t_max, 500)
    sol = solve_ivp(sugar_ode, t_span, [Q0], args=(R, X), t_eval=t_eval)
    return sol.t, sol.y[0]

# User Input Section
print("### Sugar Dynamics: Impact of Different Inflow/Outflow Rates ###")
Q0 = float(input("Enter initial sugar quantity (Q0 in grams): "))
X = float(input("Enter inflow concentration (X in grams per liter): "))
R_values = list(
    map(float, input("Enter inflow/outflow rates (comma-separated, e.g., 1,2,3): ").split(","))
)
t_max = float(input("Enter time range for simulation (in minutes): "))

# Plotting Section
plt.figure(figsize=(10, 6))
for R in R_values:
    t, Q = solve_sugar_dynamics(Q0, X, R, t_max)
    plt.plot(t, Q, label=f"R = {R} L/min")

# Customize the plot
plt.xlabel("Time (minutes)")
plt.ylabel("Sugar Quantity (grams)")
plt.title("Impact of Different Inflow/Outflow Rates on Sugar Dynamics")
plt.legend()
plt.grid()
plt.show()
```
